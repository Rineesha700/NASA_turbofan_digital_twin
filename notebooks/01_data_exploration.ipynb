```python
sensor_columns = [col for col in train.columns if col.startswith("sensor_")]

sensor_stats = train[sensor_columns].describe().T[
    ["mean", "std", "min", "max"]
]

sensor_stats
```

```python
constant_sensors = [
    col for col in sensor_columns
    if train[col].nunique() <= 1
]

print("Constant sensors:")
print(constant_sensors)
```

```python
print("Number of constant sensors:", len(constant_sensors))
```

```python
useful_sensors = [
    col for col in sensor_columns
    if col not in constant_sensors
]

features = [
    "cycle",
    "setting_1",
    "setting_2",
    "setting_3"
] + useful_sensors

model_data = train[["engine_id"] + features + ["RUL"]].copy()

print("Original columns:", len(train.columns))
print("ML features:", len(features))
print("Model data shape:", model_data.shape)
```

```python
correlations = (
    model_data[features + ["RUL"]]
    .corr()["RUL"]
    .drop("RUL")
    .sort_values()
)

correlations
```

```python
top_sensors = (
    correlations.abs()
    .sort_values(ascending=False)
    .head(5)
    .index
)

print("Top sensors:")
print(top_sensors.tolist())
```

```python
for sensor in top_sensors:
    plt.figure(figsize=(8, 4))
    plt.scatter(
        model_data[sensor],
        model_data["RUL"],
        alpha=0.2
    )
    plt.xlabel(sensor)
    plt.ylabel("RUL")
    plt.title(f"{sensor} vs Remaining Useful Life")
    plt.grid(True)
    plt.show()
```

```python
rolling_features = train[["engine_id", "cycle"] + useful_sensors].copy()

for sensor in useful_sensors:
    rolling_features[f"{sensor}_mean_5"] = (
        rolling_features
        .groupby("engine_id")[sensor]
        .transform(lambda x: x.rolling(5, min_periods=1).mean())
    )

    rolling_features[f"{sensor}_std_5"] = (
        rolling_features
        .groupby("engine_id")[sensor]
        .transform(lambda x: x.rolling(5, min_periods=1).std())
    )

rolling_features = rolling_features.fillna(0)

print("Shape:", rolling_features.shape)
```

```python
print("Constant sensors:", constant_sensors)
print("Model data shape:", model_data.shape)
```

```python
print("Shape:", rolling_features.shape)
```

```python
ml_data = rolling_features.merge(
    train[["engine_id", "cycle", "RUL"]],
    on=["engine_id", "cycle"],
    how="left"
)

print("ML dataset shape:", ml_data.shape)
print(ml_data.head())
```

```python
from sklearn.model_selection import train_test_split

engine_ids = ml_data["engine_id"].unique()

train_engines, val_engines = train_test_split(
    engine_ids,
    test_size=0.2,
    random_state=42
)

train_data = ml_data[ml_data["engine_id"].isin(train_engines)].copy()
val_data = ml_data[ml_data["engine_id"].isin(val_engines)].copy()

print("Training engines:", len(train_engines))
print("Validation engines:", len(val_engines))

print("Training rows:", len(train_data))
print("Validation rows:", len(val_data))
```

```python
feature_columns = [
    col for col in ml_data.columns
    if col not in ["engine_id", "RUL"]
]

X_train = train_data[feature_columns]
y_train = train_data["RUL"]

X_val = val_data[feature_columns]
y_val = val_data["RUL"]

print("Features:", len(feature_columns))
print("X_train:", X_train.shape)
print("X_val:", X_val.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error

rf_model = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

rf_predictions = rf_model.predict(X_val)

rf_mae = mean_absolute_error(y_val, rf_predictions)
rf_rmse = np.sqrt(mean_squared_error(y_val, rf_predictions))

print(f"Random Forest MAE: {rf_mae:.2f}")
print(f"Random Forest RMSE: {rf_rmse:.2f}")
```

```python
from xgboost import XGBRegressor

xgb_model = XGBRegressor(
    n_estimators=500,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="reg:squarederror",
    random_state=42,
    n_jobs=-1
)

xgb_model.fit(X_train, y_train)

xgb_predictions = xgb_model.predict(X_val)

xgb_mae = mean_absolute_error(y_val, xgb_predictions)
xgb_rmse = np.sqrt(mean_squared_error(y_val, xgb_predictions))

print(f"XGBoost MAE: {xgb_mae:.2f}")
print(f"XGBoost RMSE: {xgb_rmse:.2f}")
```

```python
results = pd.DataFrame({
    "Model": ["Random Forest", "XGBoost"],
    "MAE": [rf_mae, xgb_mae],
    "RMSE": [rf_rmse, xgb_rmse]
})

results
```

tuning xgboost


```python
from sklearn.model_selection import RandomizedSearchCV

param_grid = {
    "n_estimators": [300, 500, 700, 900],
    "max_depth": [3, 4, 5, 6, 8],
    "learning_rate": [0.02, 0.03, 0.05, 0.08, 0.1],
    "subsample": [0.7, 0.8, 0.9, 1.0],
    "colsample_bytree": [0.7, 0.8, 0.9, 1.0],
    "min_child_weight": [1, 3, 5, 7]
}

xgb_base = XGBRegressor(
    objective="reg:squarederror",
    random_state=42,
    n_jobs=-1
)

xgb_search = RandomizedSearchCV(
    estimator=xgb_base,
    param_distributions=param_grid,
    n_iter=20,
    scoring="neg_mean_absolute_error",
    cv=3,
    random_state=42,
    n_jobs=-1,
    verbose=1
)

xgb_search.fit(X_train, y_train)

print("Best parameters:")
print(xgb_search.best_params_)
```

```python
tuned_xgb = xgb_search.best_estimator_

tuned_predictions = tuned_xgb.predict(X_val)

tuned_mae = mean_absolute_error(y_val, tuned_predictions)
tuned_rmse = np.sqrt(mean_squared_error(y_val, tuned_predictions))

print(f"Tuned XGBoost MAE: {tuned_mae:.2f}")
print(f"Tuned XGBoost RMSE: {tuned_rmse:.2f}")
```

```python
final_results = pd.DataFrame({
    "Model": [
        "Random Forest",
        "XGBoost",
        "Tuned XGBoost"
    ],
    "MAE": [
        rf_mae,
        xgb_mae,
        tuned_mae
    ],
    "RMSE": [
        rf_rmse,
        xgb_rmse,
        tuned_rmse
    ]
})

final_results
```

```python
test_path = "../data/raw/test_FD001.txt"
rul_path = "../data/raw/RUL_FD001.txt"

test = pd.read_csv(
    test_path,
    sep=r"\s+",
    header=None
)

test_rul = pd.read_csv(
    rul_path,
    sep=r"\s+",
    header=None
)

print("Test shape:", test.shape)
print("RUL shape:", test_rul.shape)
```

```python
test.columns = columns

test.head()
```

```python
test["engine_id"] = test["engine_id"].astype(int)
test["cycle"] = test["cycle"].astype(int)

print("Number of test engines:", test["engine_id"].nunique())
```

```python
test_rul_values = test_rul.iloc[:, 0].values

last_cycles = test.groupby("engine_id")["cycle"].max().values

test["RUL"] = (
    test["engine_id"].map(
        dict(zip(
            range(1, len(test_rul_values) + 1),
            last_cycles + test_rul_values
        ))
    )
    - test["cycle"]
)

test[["engine_id", "cycle", "RUL"]].head()
```

```python
print("Minimum test RUL:", test["RUL"].min())
print("Maximum test RUL:", test["RUL"].max())
print("Negative RUL values:", (test["RUL"] < 0).sum())
```

```python
print("Number of model features:", len(feature_columns))
print(feature_columns)
```

```python
test_features = test.drop(columns=["RUL"])
```

```python
# Sensors used for modeling
model_sensors = [
    "sensor_2", "sensor_3", "sensor_4",
    "sensor_6", "sensor_7", "sensor_8",
    "sensor_9", "sensor_11", "sensor_12",
    "sensor_13", "sensor_14", "sensor_15",
    "sensor_17", "sensor_20", "sensor_21"
]

# Create rolling features separately for each engine
test_features = test.drop(columns=["RUL"]).copy()

for sensor in model_sensors:

    test_features[f"{sensor}_mean_5"] = (
        test_features.groupby("engine_id")[sensor]
        .transform(lambda x: x.rolling(window=5, min_periods=1).mean())
    )

    test_features[f"{sensor}_std_5"] = (
        test_features.groupby("engine_id")[sensor]
        .transform(lambda x: x.rolling(window=5, min_periods=1).std())
    )

# First rolling standard deviation is NaN
test_features = test_features.fillna(0)

print("Test feature shape:", test_features.shape)
```

```python
X_test = test_features[feature_columns]

print("X_test shape:", X_test.shape)
print("Expected features:", len(feature_columns))
print("Actual features:", X_test.shape[1])
```

```python
print("Missing values:", X_test.isna().sum().sum())
```

```python
test_results = test[["engine_id", "cycle", "RUL"]].copy()
test_results["Predicted_RUL"] = test_predictions

test_results.head(10)
```

```python
# Get the final observed cycle for each test engine
final_cycle_results = (
    test_results
    .sort_values(["engine_id", "cycle"])
    .groupby("engine_id")
    .tail(1)
    .sort_values("engine_id")
    .reset_index(drop=True)
)

# Compare predicted vs true RUL
final_cycle_results[["engine_id", "RUL", "Predicted_RUL"]].head(10)
```

```python
test_mae = mean_absolute_error(
    final_cycle_results["RUL"],
    final_cycle_results["Predicted_RUL"]
)

test_rmse = np.sqrt(
    mean_squared_error(
        final_cycle_results["RUL"],
        final_cycle_results["Predicted_RUL"]
    )
)

print(f"Test MAE: {test_mae:.2f} cycles")
print(f"Test RMSE: {test_rmse:.2f} cycles")
```
